```python
import temp_attitude_propagator as temp
import numpy as np
import matplotlib.pyplot as plt
import math

angles, res = temp.get_res()

plt.figure(figsize=(7, 7))
plt.plot(angles/math.pi * 180, res[:, 0]/math.pi * 180, label="Front Sun Sensor")
plt.plot(angles/math.pi * 180, res[:, 1]/math.pi * 180, label="Side Sun Sensor")
plt.title("Angle between Measured and True Sun Vectors throughout an Orbit")
plt.xlabel("Angular Position in Orbit [deg]")
plt.ylabel("Angle between Sun Vectors [deg]")
plt.legend()
plt.grid()
plt.show()
```
