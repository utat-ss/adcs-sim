```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import math

import data_process

times, normed_diffs, alts  = data_process.examine_sp3_data(True, True)

plt.figure(figsize=(7, 7))

plt.plot(times, normed_diffs / 1000.0, label="Total Position Deviation")
plt.plot(times, alts / 1000.0, label="Altitude Deviation")

plt.title("Deviation [km] from CASSIOPE SP3 Data over Time [min]")
plt.xlabel("Time Elapsed [min]")
plt.ylabel("Difference in Position [km]")
plt.grid()
plt.legend()
plt.show()
```
